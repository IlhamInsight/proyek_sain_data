# Business Understanding

Tahap **Business Understanding** merupakan fondasi awal dalam metodologi *Cross-Industry Standard Process for Data Mining* (CRISP-DM) yang bertujuan untuk memahami latar belakang masalah, tujuan domain, kebutuhan data, serta merumuskan rencana proyek sains data yang terukur.

---

### 1. Latar Belakang & Konteks Geospasial

Pemantauan kualitas udara secara berkelanjutan sangat krusial untuk menganalisis tingkat paparan polutan atmosfer terhadap lingkungan dan kesehatan masyarakat. Proyek sains data ini berfokus pada area studi spesifik di kawasan perdesaan:

* **Lokasi Studi**: Desa Mendenrejo, Kecamatan Kradenan, Kabupaten Blora, Provinsi Jawa Tengah, Indonesia.
* **Batas Wilayah Geospasial**: Didefinisikan menggunakan poligon batas riil multi-titik (non-persegi/mengikuti kontur batas desa) dalam format standar **GeoJSON (WGS84 / EPSG:4326)**.
* **Tantangan Lingkungan**: Wilayah Desa Mendenrejo belum memiliki stasiun pemantau kualitas udara darat (*ground-based monitoring station* / ISPU). Oleh karena itu, pendekatan penginderaan jauh (*remote sensing*) berbasis satelit menjadi solusi utama.

---

### 2. Sumber Data & Pendekatan Penginderaan Jauh

Untuk mengatasi keterbatasan sensor darat, proyek ini memanfaatkan data observasi satelit bumi dari misi **Copernicus Sentinel-5P TROPOMI Level-2 (L2)** yang diakses secara terprogram melalui **Copernicus Data Space Ecosystem (CDSE) openEO API**:

* **Periode Pemantauan**: Deret waktu harian (*daily time-series*) selama 1 tahun penuh, terhitung mulai tanggal **1 September 2025 hingga 31 Agustus 2026** (365 hari).
* **Agregasi Spasial**: Nilai konsentrasi polutan diekstraksi berdasarkan rata-rata spasial (*spatial mean reducer*) di dalam batas poligon Desa Mendenrejo.

---

### 3. Parameter Polutan Atmosfer yang Dipantau

Empat parameter gas polutan atmosfer utama dipantau sesuai dengan karakteristik geografis dan aktivitas antropogenik di wilayah Desa Mendenrejo:

1. **Karbon Monoksida (CO)**:
   * *Satuan*: $\text{mol/m}^2$ (Kolom total atmosfer).
   * *Sumber Emisi*: Pembakaran tidak sempurna dari kendaraan bermotor, aktivitas domestik rumah tangga, serta pembakaran sisa biomassa pertanian.
2. **Nitrogen Dioksida (NO₂)**:
   * *Satuan*: $\text{mol/m}^2$ (Kolom troposferik).
   * *Sumber Emisi*: Emisi transportasi jalan raya, pembakaran bahan bakar fosil, dan transmisi polusi udara regional.
3. **Sulfur Dioksida (SO₂)**:
   * *Satuan*: $\text{mol/m}^2$ (Kolom total atmosfer).
   * *Sumber Emisi*: Emisi pembakaran material organik/belerang dan aktivitas industri energi regional.
4. **Metana (CH₄)**:
   * *Satuan*: $\text{ppb}$ / $\text{mol/m}^2$.
   * *Sumber Emisi*: Dekomposisi anaerobik dari area persawahan tergenang, peternakan lokal, serta pembusukan limbah organik.

---

### 4. Tujuan Proyek Sains Data (Data Science Objectives)

Proyek ini bertujuan untuk menyelesaikan serangkaian target analitis dan teknis:

1. **Pengambilan Data (Data Acquisition)**: Mengotomatisasi ekstraksi 4 parameter polutan satelit Sentinel-5P dari CDSE openEO ke dalam 1 dataset CSV terpadu.
2. **Pemahaman Data (Data Understanding)**: Menganalisis distribusi statistik, missing values akibat tutupan awan, korelasi antar-polutan, dan pola musiman (*seasonal trends*).
3. **Persiapan Data (Data Preparation)**: Menangani data hilang (*imputation*), pembersihan anomali/outlier, serta pembentukan fitur lag dan kalender temporal.
4. **Pemodelan Prediktif (Time-Series Modeling)**: Mengembangkan model pembelajaran mesin (*machine learning / time-series forecasting*) untuk memprediksi konsentrasi polutan atmosfer di masa mendatang.
5. **Evaluasi & Diseminasi**: Menguji performa model (RMSE, MAE, R²) serta menyajikan dokumentasi interaktif berbasis MyST Markdown Book.
