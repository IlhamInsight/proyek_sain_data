# Business Understanding

Tahap **Business Understanding** berfokus pada pemahaman konteks domain, kebutuhan pemantauan, serta pemetaan fokus penerapan sains data pada lokasi studi yang telah ditentukan.

---

### Pemahaman Konteks Pemantauan Kualitas Udara
Pemantauan dan pemodelan kualitas udara dilakukan untuk menganalisis dinamika polutan atmosfer pada lokasi spesifik, yaitu **Desa Mendenrejo, Kecamatan Kradenan, Kabupaten Blora, Jawa Tengah** (Koordinat Pusat Spasial: Latitude **-7.2479**, Longitude **111.4440**).

Karena kawasan perdesaan ini belum memiliki stasiun pemantau fisik darat (*ground-based monitoring station*), pendekatan penginderaan jauh (*remote sensing*) dengan memanfaatkan data **Copernicus Atmosphere Monitoring Service (CAMS)** digunakan sebagai sumber data utama. Pemantauan dilakukan terhadap data deret waktu per jam (*hourly time-series*) selama periode 1 tahun penuh.

---

### Pemahaman Parameter Polutan yang Dipantau
Empat parameter polutan atmosferik dipantau dan dianalisis sesuai karakteristik lokasi Desa Mendenrejo:

1. **Karbon Monoksida (CO)**: Mengukur emisi hasil pembakaran tidak sempurna dari kendaraan bermotor lokal, aktivitas domestik, dan pembakaran sisa biomassa pertanian.
2. **Nitrogen Dioksida (NO₂)**: Mengukur pencemaran akibat aktivitas transportasi dan emisi pembakaran regional.
3. **Sulfur Dioksida (SO₂)**: Mengukur indikator emisi pembakaran bahan bakar fosil dan material organik.
4. **Metana (CH₄)**: Mengukur konsentrasi gas rumah kaca yang sangat relevan dengan aktivitas perdesaan seperti lahan pertanian sawah tergenang, peternakan, dan degradasi limbah organik.

---

### Fokus Penerapan Sains Data
Fokus utama proyek sains data ini adalah mempraktikkan alur pemodelan prediktif deret waktu (*Air Quality Time-Series Forecasting*):

- **Analisis Pola Temporal**: Memahami fluktuasi harian (*diurnal cycle*) dan tren musiman polutan pada lokasi studi.
- **Kebutuhan Pemodelan**: Menyiapkan data mentah untuk tahap eksplorasi (*Data Understanding*), pembersihan/rekayasa fitur (*Data Preparation*), hingga pelatihan model pembelajaran mesin (*Time-Series Modeling*) untuk memprediksi tingkat konsentrasi polutan di masa mendatang.
