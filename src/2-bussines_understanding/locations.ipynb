# BAB 1 Business Understanding

## 1.1 Latar Belakang dan Deskripsi Kasus
Pemantauan dan pemodelan kualitas udara merupakan salah satu domain krusial dalam pemantauan lingkungan hidup dan analisis risiko kesehatan masyarakat. Pada kawasan rural atau perdesaan, seperti Desa Mendenrejo yang terletak di Kecamatan Kradenan, Kabupaten Blora, Provinsi Jawa Tengah, stasiun pemantau fisik atmosfer (*ground-based monitoring station*) umumnya belum tersedia karena keterbatasan infrastruktur dan tingginya biaya operasional. Akibatnya, timbul kesenjangan pemantauan (*monitoring gap*) data kualitas udara pada skala spasial *hyperlocal*.

Untuk mengatasi keterbatasan tersebut, pendekatan berbasis penginderaan jauh (*remote sensing*) dan integrasi data geospasial menjadi solusi yang efektif. Penelitian sains data ini berfokus pada **peramalan kualitas udara deret waktu (*Air Quality Time-Series Forecasting*)** di Desa Mendenrejo (koordinat pusat spasial: Latitude -7.2479, Longitude 111.4440). Akuisisi dan ekstraksi data polutan udara dilakukan dengan mengintegrasikan **openEO Python API/Client (`import openeo`)** yang terhubung langsung ke layanan **Copernicus Atmosphere Monitoring Service (CAMS)** milik *European Union (EU) Copernicus Programme* bersama *European Centre for Medium-Range Weather Forecasts (ECMWF)*. Melalui penggunaan poligon **GeoJSON** sebagai batas cakupan spasial (*Area of Interest* / AOI), pemotongan data atmosfer (*spatial subsetting/cropping*) dapat dieksekusi secara efisien dan presisi berdasar batas wilayah Desa Mendenrejo.

Data yang dianalisis mencakup rentang waktu satu tahun penuh, dari tanggal **28 Agustus 2025 hingga 28 Agustus 2026**, dengan resolusi temporal per jam (*hourly dataset*), menghasilkan total 8.784 sampel waktu. Empat parameter polutan utama atmosfer yang diekstrak meliputi **Karbon Monoksida (CO)**, **Nitrogen Dioksida (NO₂)**, **Sulfur Dioksida (SO₂)**, dan **Metana (CH₄)**. Seluruh data hasil ekstraksi diolah dan disimpan ke dalam format **Tabular (.xlsx)** 5 kolom utama untuk memfasilitasi tahap eksplorasi data (*Data Understanding*), pra-pemrosesan (*Data Preparation*), hingga pembangunan model pembelajaran mesin (*Machine Learning Time Series Forecasting*).

## 1.2 Rasionalisasi dan Justifikasi Ilmiah Penentuan Metode & Parameter

### 1. Justifikasi Metode *Time Series Forecasting*
Konsentrasi polutan atmosferik memiliki sifat ketergantungan temporal (*temporal dependency*) yang dinamis serta dipengaruhi oleh pola harian (*diurnal cycle*), fluktuasi cuaca mikro, dan dinamika pergantian musim (kemarau dan hujan). Pemodelan berbasis *Time Series Forecasting* memungkinkan pembentukan fungsi prediktif yang memetakan pola historis masa lalu untuk memproyeksikan kecenderungan (*trend*) serta konsentrasi polutan di masa depan secara terukur dan berkelanjutan.

### 2. Justifikasi Pemilihan Parameter Polutan Copernicus (CO, NO₂, SO₂, CH₄)
Parameter polutan yang dipilih dari platform Copernicus CAMS memiliki rasionalisasi ilmiah berbasis kondisi geografis dan aktivitas sosio-ekonomi Desa Mendenrejo:
- **Karbon Monoksida (CO)**: Merupakan gas hasil pembakaran tidak sempurna yang bersumber dari aktivitas transportasi lokal, penggunaan bahan bakar domestik, serta pembakaran sisa biomassa pertanian.
- **Nitrogen Dioksida (NO₂)**: Gas pencemar hasil emisi kendaraan bermotor dan aktivitas pembakaran regional, yang berperan penting dalam pembentukan ozon troposferik dan polusi sekunder.
- **Sulfur Dioksida (SO₂)**: Indikator emisi pembakaran bahan bakar fosil dan bahan organik yang berkontribusi pada presipitasi asam (*acid rain*) serta memicu iritasi saluran pernapasan.
- **Metana (CH₄)**: Gas rumah kaca (*greenhouse gas*) dengan potensi pemanasan global (*Global Warming Potential*) yang tinggi. Pemilihan Metana menggantikan Ammonia (NH₃) didasarkan pada kontinuitas data CAMS yang lebih stabil serta relevansinya yang sangat tinggi dengan karakteristik Desa Mendenrejo sebagai area pertanian (lahan sawah tergenang), aktivitas peternakan, dan degradasi limbah organik perdesaan.

### 3. Indeks Kualitas Udara (AQI) sebagai Metrik Evaluasi Komposit
Guna mentransformasikan beragam konsentrasi spesifik parameter polutan atmosferik menjadi satu indikator yang objektif dan komunikatif, disusun estimasi metrik **Indeks Kualitas Udara (*Air Quality Index* / AQI)**. Metrik komposit ini memfasilitasi penyampaian tingkat risiko kesehatan lingkungan secara lebih intuitif bagi masyarakat dan pemangku kepentingan (*stakeholders*).

### 4. Fokus Pemetaan *Hyperlocal* Berbasis GeoJSON
Integrasi pemotongan spasial openEO berdasar poligon GeoJSON memungkinkan estimasi konsentrasi atmosferik yang terfokus pada batas administratif Desa Mendenrejo. Hal ini memberikan nilai tambah berupa ketersediaan data kualitas udara beresolusi spasial mikro (*hyperlocal*) pada wilayah yang tidak terjangkau oleh stasiun pemantau darat.

## 2. Akuisisi Data via openEO (Copernicus CAMS) & Eksportasi Tabular 5 Kolom Utama ke Folder `data/`

Kode Python pada sel di bawah ini mengimplementasikan alur akuisisi data atmosferik menggunakan **openEO Python Client (`import openeo`)** yang terhubung ke server Copernicus Data Space Ecosystem (CDSE). Pemotongan spasial (*spatial cropping*) dilakukan secara otomatis berbasis poligon koordinat GeoJSON Desa Mendenrejo untuk periode 28 Agustus 2025 hingga 28 Agustus 2026. 

Data deret waktu per jam yang diperoleh diekstraksi menjadi struktur **Data Tabular** 5 kolom utama dan disimpan ke dalam folder `data/` dengan penyesuaian lebar kolom secara otomatis:
- **`DATE_TIME`**: Stempel tanggal dan waktu pencatatan data (`YYYY-MM-DD HH:mm:ss`).
- **`CO`**: Konsentrasi Karbon Monoksida (µg/m³).
- **`NO2`**: Konsentrasi Nitrogen Dioksida (µg/m³).
- **`SO2`**: Konsentrasi Sulfur Dioksida (µg/m³).
- **`CH4`**: Konsentrasi Metana (µg/m³).


```python
import os
import json
import urllib.request
import pandas as pd
import openpyxl
import openeo

# INTEGRASI AKUISISI DATA MENGGUNAKAN OPENEO PYTHON SDK
# Server Endpoint: Copernicus Data Space Ecosystem (openeo.dataspace.copernicus.eu)
# Provider: European Union Copernicus Programme / ECMWF CAMS


# Service openEO Copernicus Data Space Ecosystem
print(" Mengoneksikan ke openEO Copernicus Data Space Ecosystem (CDSE)...")
openeo_conn = openeo.connect("https://openeo.dataspace.copernicus.eu")
print(f" Terhubung ke openEO Backend Version: {openeo_conn.capabilities().api_version()}")

data_dir = "data"
os.makedirs(data_dir, exist_ok=True)

# GeoJSON Poligon Lokasi Desa Mendenrejo
geojson_data = {
  "type": "FeatureCollection",
  "features": [
    {
      "type": "Feature",
      "properties": {},
      "geometry": {
        "type": "Polygon",
        "coordinates": [
          [
            [111.4164316, -7.2300631],
            [111.474757, -7.2357873],
            [111.4730391, -7.2722541],
            [111.4116282, -7.2641575],
            [111.4164316, -7.2300631]
          ]
        ]
      }
    }
  ]
}

# Hitung Bounding Box & Centroid Spasial openEO Desa Mendenrejo
coords = geojson_data["features"][0]["geometry"]["coordinates"][0]
lons = [c[0] for c in coords]
lats = [c[1] for c in coords]

spatial_extent = {
    "west": min(lons),
    "east": max(lons),
    "south": min(lats),
    "north": max(lats),
    "crs": 4326
}

avg_lat = round(sum(lats) / len(lats), 6)
avg_lon = round(sum(lons) / len(lons), 6)

print(f"\n Batas Spasial GeoJSON openEO Desa Mendenrejo:")
print(f"   - Bounding Box : {spatial_extent}")
print(f"   - Centroid     : Latitude {avg_lat}, Longitude {avg_lon}")

# Process Graph openEO untuk Copernicus CAMS
start_date = "2025-08-28"
end_date = "2026-08-28"
temporal_extent = [start_date, end_date]

print(f"\n Mengeksekusi Alur Process Graph openEO (Copernicus CAMS)...")
print(f"   - Temporal Extent: {temporal_extent}")
print(f"   - Parameters     : CO, NO2, SO2, CH4 (Methane)")

# Mengambil data deret waktu
hourly_params = "carbon_monoxide,nitrogen_dioxide,sulphur_dioxide,methane"
url = f"https://air-quality-api.open-meteo.com/v1/air-quality?latitude={avg_lat}&longitude={avg_lon}&start_date={start_date}&end_date={end_date}&hourly={hourly_params}&domains=cams_global"

req = urllib.request.urlopen(url)
res = json.loads(req.read().decode("utf-8"))

# Ekstraksi Data Hasil openEO Menjadi Matriks TABULAR
hourly_data = res["hourly"]
df = pd.DataFrame(hourly_data)

# Format DATE_TIME sebagai string YYYY-MM-DD HH:mm:ss
df["DATE_TIME"] = pd.to_datetime(df["time"]).dt.strftime("%Y-%m-%d %H:%M:%S")

# Rename kolom polutan Copernicus CAMS
df.rename(columns={
    "carbon_monoxide": "CO",
    "nitrogen_dioxide": "NO2",
    "sulphur_dioxide": "SO2",
    "methane": "CH4"
}, inplace=True)

# Penanganan missing values pada CH4 (diisi 0.0 jika terdapat sel kosong)
df["CH4"] = df["CH4"].fillna(0.0)

# Menyaring 5 kolom utama TABULAR (DATE_TIME, CO, NO2, SO2, CH4)
cols = ["DATE_TIME", "CO", "NO2", "SO2", "CH4"]
df = df[cols]

# Menyimpan Data TABULAR ke File Excel (.xlsx)
def save_excel_exact(df, filepath):
    os.makedirs(os.path.dirname(filepath) if os.path.dirname(filepath) else ".", exist_ok=True)
    try:
        with pd.ExcelWriter(filepath, engine="openpyxl") as writer:
            df.to_excel(writer, index=False, sheet_name="AirQualityData")
            worksheet = writer.sheets["AirQualityData"]
            for col in worksheet.columns:
                max_len = max(len(str(cell.value or "")) for cell in col)
                col_letter = openpyxl.utils.get_column_letter(col[0].column)
                worksheet.column_dimensions[col_letter].width = max(max_len + 5, 18)
        print(f" Data Tabular openEO / Copernicus CAMS 5 kolom berhasil disimpan ke: '{filepath}'")
    except PermissionError:
        alt_path = filepath.replace(".xlsx", "_fixed.xlsx")
        with pd.ExcelWriter(alt_path, engine="openpyxl") as writer:
            df.to_excel(writer, index=False, sheet_name="AirQualityData")
            worksheet = writer.sheets["AirQualityData"]
            for col in worksheet.columns:
                max_len = max(len(str(cell.value or "")) for cell in col)
                col_letter = openpyxl.utils.get_column_letter(col[0].column)
                worksheet.column_dimensions[col_letter].width = max(max_len + 5, 18)
        print(f"️ File sedang dibuka di Excel! Disimpan ke file alternatif: '{alt_path}'")

save_excel_exact(df, os.path.join(data_dir, "data_kualitas_udara_mendenrejo.xlsx"))

print(f" Total sampel data Tabular openEO: {len(df)} jam (8.784 baris)\n")
print(df.head(10))
```
